```python
import pandas as pd
```

```python
l = ['Indian', 'American', 'British', 'Canadian', 'Australian']

pd.Series(l)
```

```python
marks = [90, 80, 70, 60, 50]
subjects = ['Maths', 'Science', 'English', 'History', 'Geography']

pd.Series(marks, index=subjects)
```

```python
marks = [90, 80, 70, 60, 50]
subjects = ['Maths', 'Science', 'English', 'History', 'Geography']
pd.Series(marks, index=subjects,name='Marks')
```

```python
# by using dictionary
marks_dict = {'Maths': 90, 'Science': 80, 'English': 70, 'History': 60, 'Geography': 50}
Marks = pd.Series(marks_dict,name='Bhushan Marks')
```
